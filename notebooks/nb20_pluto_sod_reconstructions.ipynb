# Sod problem in PLUTO: reconstructions and convergence

### Instructions for this live exercise:

**1.** Add minimal code lines to `nb18`, so that it prints the exact solution at a few selected times. Save the file.

**2.** Make three run directories (`Sod_01`, `Sod_02`, and `Sod_03`), one per configuration. Open the three `definitions.h` and write down, in a markdown cell, which reconstruction and which time marching each one uses.

**3.** Configure, compile, and run each directory separately, as in Lecture 22: `python3 "$PLUTO_DIR/setup.py"`, then `make`, then `./pluto`. The three directories have different `definitions.h`, so each one needs its own build. Inside a directory, `make clean` before `make` whenever you change `definitions.h`.

**4.** Read the last snapshot of `Sod_01` with the cell below and make the same 4-panel plot. Check that the time is the one you asked for in `pluto.ini`.

**5.** Add `Sod_02` and `Sod_03` to the same figure, one colour each, together with the exact solution saved from `nb18`. Then discuss, in a markdown cell: which of the three waves looks different between the 3 runs, which one looks the same in all of them, and why. Look in particular at the contact and at the top of the rarefaction.

**6.** Repeat the three runs at four resolutions, $n_x = 64, 128, 256, 512$. Only `X1-grid` in `pluto.ini` changes, so no recompilation is needed: edit the file and rerun `./pluto`. Keep the outputs of each run in its own directory, or rename them, so that nothing is overwritten.

**7.** Write a function that returns the error of a run against the exact solution of `nb18` in the $L_2$ norm,

$$E_2 = \left[\frac{1}{N}\sum_{i}\left(q_i - q^{\rm exact}_i\right)^{2}\right]^{1/2}$$

evaluated on the grid of the run. Remember that the exact solution has to be sampled at the same resolution!!

**8.** Build the **error map**: a matrix with one row per configuration and one column per resolution, with $\log_{10}E_2$ of the density as the colour. Use `plt.imshow` or `plt.pcolormesh`, label both axes with the configuration names and the resolutions, and add a colour bar. Make one map for $\rho$ and one for $p$.

**9.** From the same numbers, make a log-log plot of $E_2$ against $n_x$, one line per configuration, and measure the order of convergence between consecutive resolutions.

**10.** Discuss, in a markdown cell: what order each configuration reaches and whether it matches what the lecture claims, where in the domain the error is concentrated, and whether the extra cost of the parabolic reconstruction buys more than simply doubling the resolution of the cheapest one.
